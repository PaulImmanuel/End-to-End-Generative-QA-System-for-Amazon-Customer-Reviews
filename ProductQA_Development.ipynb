## Q&A pairs

```python
import pandas as pd
from transformers import pipeline, T5Tokenizer
from tqdm import tqdm
import torch
import os

# --- Configuration ---
# Let's process all 25000 rows we loaded in 'df'
ROWS_TO_PROCESS = 25000 
LOCAL_QG_MODEL_PATH = os.path.join("./models", "qg_model")
LOCAL_QA_MODEL_PATH = os.path.join("./models", "qa_model")
OUTPUT_FILE = "generated_qa_dataset.csv"

# --- Get Data and Device ---
%store -r device_to_use
print(f"Using device: {'GPU (cuda:0)' if device_to_use == 0 else 'CPU'}")

if 'df' in locals() and len(df) >= ROWS_TO_PROCESS:
    # Use the full 25000-row sample this time
    sample_df = df.head(ROWS_TO_PROCESS).copy()
    print(f"Starting Q&A generation for {len(sample_df)} reviews.")
else:
    print(f"Error: 'df' not found or is smaller than {ROWS_TO_PROCESS} rows.")
    print("Please re-run the cell that loads 'Reviews.csv' and ensure 'nrows' is at least 25000.")
    raise NameError("'df' DataFrame not found or has too few rows.")

# --- Setup Models (Loading from Local Disk) ---
try:
    print(f"\nLoading QG tokenizer from local path: {LOCAL_QG_MODEL_PATH}...")
    qg_tokenizer = T5Tokenizer.from_pretrained(LOCAL_QG_MODEL_PATH)
    print("QG tokenizer loaded.")

    print(f"\nLoading QG model from local path: {LOCAL_QG_MODEL_PATH}...")
    qg_pipeline = pipeline(
        "text2text-generation",
        model=LOCAL_QG_MODEL_PATH,
        tokenizer=qg_tokenizer,
        device=device_to_use
    )
    print("QG model loaded.")

    print(f"\nLoading QA model from local path: {LOCAL_QA_MODEL_PATH}...")
    qa_pipeline = pipeline(
        "question-answering",
        model=LOCAL_QA_MODEL_PATH,
        device=device_to_use
    )
    print("QA model loaded.")

    # --- Q&A Generation Function (with fix) ---
    def generate_qa_pair(row_text):
        context = row_text
        if not isinstance(context, str) or len(context.split()) < 20:
            return None, None
        try:
            truncated_context = context[:1500]
            
            # *** FIX: Use max_new_tokens instead of max_length ***
            generated_output = qg_pipeline(
                truncated_context, 
                max_new_tokens=64,  # <-- This is the fix
                num_beams=4, 
                early_stopping=True
            )
            
            if not generated_output: return None, None 
            question = generated_output[0]['generated_text']

            if not question or '?' not in question: return None, None

            answer_dict = qa_pipeline(question=question, context=context)
            if not answer_dict: return None, None

            answer = answer_dict.get('answer', None)
            score = answer_dict.get('score', 0) 

            if answer is None or score < 0.3 or len(answer.split()) < 2:
                return None, None
            return question, answer
        except Exception:
             return None, None

    # --- Run the Process ---
    if not sample_df.empty:
        print(f"\n🚀 Generating Q&A pairs for {len(sample_df)} reviews...")
        print("This will take a significant amount of time (e.g., 20-40 minutes).")
        print("This is a good time to take a break!")
        
        tqdm.pandas(desc="Processing Reviews")
        results = sample_df['Text'].progress_apply(generate_qa_pair)
        sample_df[['generated_question', 'generated_answer']] = pd.DataFrame(results.tolist(), index=sample_df.index)

        print("\n✅ Generation complete.")
        
        # --- Save Results ---
        final_qa_data = sample_df.dropna(subset=['generated_question', 'generated_answer']).copy()
        
        # We need the context, question, and answer for training
        # Let's rename 'Text' to 'context' for clarity in the next phase
        final_qa_data.rename(columns={'Text': 'context'}, inplace=True)
        
        # Select only the columns we need for training
        training_df = final_qa_data[['context', 'generated_question', 'generated_answer']]
        
        # Save to a new CSV file
        training_df.to_csv(OUTPUT_FILE, index=False)
        
        print(f"\nSuccessfully created and saved {len(training_df)} valid Q&A pairs to '{OUTPUT_FILE}'.")
        print("\n--- Final Dataset (Sample): ---")
        print(training_df.head())


except Exception as e:
    print(f"\n❌ An error occurred: {e}")
```

## Splitting Data

```python
from datasets import load_dataset, DatasetDict

# --- Configuration ---
DATASET_FILE = "generated_qa_dataset.csv"

# --- Load Data ---
print(f"Loading our generated dataset from {DATASET_FILE}...")
try:
    # Load the CSV file as a Hugging Face Dataset object
    full_dataset = load_dataset('csv', data_files=DATASET_FILE, split='train')
    
    # --- Split Data ---
    # We need to find the answer_start index for our text-based answer
    # This is a bit of a workaround to add the 'answer_start' column
    def find_answer_start(example):
        # We'll store the answer as a dictionary, just like the SQuAD dataset
        answer_text = example['generated_answer']
        context_text = example['context']
        
        answer_start = context_text.find(answer_text)
        
        # If the answer isn't found (shouldn't happen, but good to check)
        if answer_start == -1:
            return {'answers': {'text': [], 'answer_start': []}}
        
        return {
            'answers': {
                'text': [answer_text],
                'answer_start': [answer_start]
            }
        }

    print("Formatting answers to SQuAD format...")
    full_dataset = full_dataset.map(find_answer_start, remove_columns=['generated_answer'])
    
    # Now, split the dataset into training (90%) and testing (10%)
    # Using shuffle=True is good practice
    train_test_split = full_dataset.train_test_split(test_size=0.1, shuffle=True, seed=42)
    
    # Rename for clarity
    datasets = DatasetDict({
        'train': train_test_split['train'],
        'test': train_test_split['test']
    })

    print("\n✅ Dataset loaded and split successfully.")
    print(datasets)
    
    print("\n--- Example Training Data: ---")
    print(datasets['train'][0])

except FileNotFoundError:
    print(f"\n*** ERROR: File not found! ***")
    print(f"Could not find '{DATASET_FILE}'. Please make sure it's in the same directory.")
except Exception as e:
    print(f"\nAn error occurred: {e}")
```

## Set Up Model, Tokenizer, and Preprocessing

```python
from transformers import AutoTokenizer, AutoModelForQuestionAnswering, TrainingArguments, Trainer, DefaultDataCollator
import os

# --- Configuration ---
# Use the local QA model we downloaded as our base for fine-tuning
MODEL_CHECKPOINT = os.path.join("./models", "qa_model")
# This is where we will save our *new* fine-tuned model
NEW_MODEL_DIR = os.path.join("./models", "my_final_qa_model")

# Tokenizer settings
MAX_LENGTH = 384 # Max length of a question + context chunk
DOC_STRIDE = 128 # Overlap between chunks if context is too long

# --- Load Tokenizer ---
print(f"Loading tokenizer from local path: {MODEL_CHECKPOINT}")
# We use 'use_fast=True' as it's required for the offset mappings
tokenizer = AutoTokenizer.from_pretrained(MODEL_CHECKPOINT, use_fast=True)
print("Tokenizer loaded.")

# --- Preprocessing Function ---
# This is the most complex step. It formats the data for the model.
def preprocess_function(examples):
    # Tokenize questions and contexts together
    # Truncation="only_second" means we only truncate the context, not the question
    tokenized_examples = tokenizer(
        examples["generated_question"],
        examples["context"],
        truncation="only_second",
        max_length=MAX_LENGTH,
        stride=DOC_STRIDE,
        return_overflowing_tokens=True,
        return_offsets_mapping=True,
        padding="max_length",
    )

    # offset_mapping gives us a map from token-level to character-level in the original text
    offset_mapping = tokenized_examples.pop("offset_mapping")
    
    # sample_mapping maps from a new feature (a chunk) back to its original example
    sample_mapping = tokenized_examples.pop("overflow_to_sample_mapping")

    # These will be our new labels
    tokenized_examples["start_positions"] = []
    tokenized_examples["end_positions"] = []

    for i, offsets in enumerate(offset_mapping):
        # Get the ID of the original example
        sample_index = sample_mapping[i]
        answers = examples["answers"][sample_index]
        
        # If there's no answer for this example, label (0, 0)
        if len(answers["answer_start"]) == 0:
            tokenized_examples["start_positions"].append(0)
            tokenized_examples["end_positions"].append(0)
            continue

        # Get the start and end character positions of the answer
        start_char = answers["answer_start"][0]
        end_char = start_char + len(answers["text"][0])

        # Get the token indices that correspond to the current feature (chunk)
        sequence_ids = tokenized_examples.sequence_ids(i)

        # Find the start and end of the context part in the tokenized input
        idx = 0
        while sequence_ids[idx] != 1: # sequence_id 1 is the context
            idx += 1
        context_start = idx
        while sequence_ids[idx] == 1:
            idx += 1
        context_end = idx - 1

        # Check if the answer is completely outside the current chunk
        if (offsets[context_start][0] > end_char or offsets[context_end][1] < start_char):
            tokenized_examples["start_positions"].append(0)
            tokenized_examples["end_positions"].append(0)
        else:
            # Find the token start index
            token_start_index = context_start
            while token_start_index <= context_end and offsets[token_start_index][0] < start_char:
                token_start_index += 1
            
            # Find the token end index
            token_end_index = context_end
            while token_end_index >= context_start and offsets[token_end_index][1] > end_char:
                token_end_index -= 1

            tokenized_examples["start_positions"].append(token_start_index)
            tokenized_examples["end_positions"].append(token_end_index)

    return tokenized_examples

# --- Apply Preprocessing ---
if 'datasets' in locals():
    print("\nApplying preprocessing to the dataset... (This may take a moment)")
    # Use .map() to apply the function to all examples
    # batched=True processes multiple examples at once
    tokenized_datasets = datasets.map(
        preprocess_function,
        batched=True,
        remove_columns=datasets["train"].column_names
    )
    print("Preprocessing complete.")
    print(tokenized_datasets)
else:
    print("Error: 'datasets' not found. Please re-run the previous cell.")
```

## Fine-Tuning Model!

```python
import torch
from transformers import AutoModelForQuestionAnswering, TrainingArguments, Trainer, DefaultDataCollator
import os

# --- Configuration ---
MODEL_CHECKPOINT = os.path.join("./models", "qa_model")
NEW_MODEL_DIR = os.path.join("./models", "my_final_qa_model")

# --- Load Model ---
print(f"Loading pre-trained model from local path: {MODEL_CHECKPOINT}")
model = AutoModelForQuestionAnswering.from_pretrained(MODEL_CHECKPOINT)
print("Model loaded.")

# --- Set Training Arguments ---
# These arguments control the fine-tuning process
training_args = TrainingArguments(
    output_dir=NEW_MODEL_DIR,
    # --- FIX ---
    logging_strategy="epoch",  # Log at the same time as evaluation
    # ---
    eval_strategy="epoch",
    learning_rate=2e-5,
    per_device_train_batch_size=16,
    per_device_eval_batch_size=16,
    num_train_epochs=3,
    weight_decay=0.01,
    push_to_hub=False,
    save_strategy="epoch",
    load_best_model_at_end=True,
    report_to="none",
)

# Data collator handles batching our examples together
data_collator = DefaultDataCollator()

# --- Initialize Trainer ---
print("Initializing Trainer...")
trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=tokenized_datasets["train"],
    eval_dataset=tokenized_datasets["test"],
    tokenizer=tokenizer,
    data_collator=data_collator,
)
print("Trainer initialized.")

# --- START FINE-TUNING ---
if 'trainer' in locals():
    print("\n🚀 *** STARTING FINE-TUNING! *** 🚀")
    print("This will run for several minutes. You'll see the loss go down.")
    
    try:
        # This is the command that starts the training
        trainer.train()
        
        print("\n✅ *** FINE-TUNING COMPLETE! *** ✅")
        
        # --- Save the Final Model ---
        print(f"Saving the best model to {NEW_MODEL_DIR}...")
        trainer.save_model()
        print("Model saved successfully.")
        
    except torch.cuda.OutOfMemoryError:
        print("\n❌ *** CUDA Out of Memory! *** ❌")
        print("Your GPU ran out of memory. Try this:")
        print("1. Restart the kernel.")
        print("2. In the 'TrainingArguments' cell, change 'per_device_train_batch_size=16' to 'per_device_train_batch_size=8'.")
        print("3. Re-run this cell.")
    except Exception as e:
        print(f"\nAn error occurred during training: {e}")
else:
    print("Error: 'trainer' not initialized. Please re-run the previous cells.")
```

```python
from datasets import load_dataset, DatasetDict
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM, TrainingArguments, Trainer, DataCollatorForSeq2Seq
import os

# --- 1. Configuration ---
DATASET_FILE = "generated_qa_dataset.csv" # The file you already made
MODEL_CHECKPOINT = os.path.join("./models", "t5_small") # The T5 model we just downloaded
NEW_MODEL_DIR = os.path.join("./models", "my_final_T5_qa_model") # The NEW final model

# --- 2. Load and Split Dataset ---
print(f"Loading our generated dataset from {DATASET_FILE}...")
try:
    # Load the CSV file as a Hugging Face Dataset object
    full_dataset = load_dataset('csv', data_files=DATASET_FILE, split='train')
    
    # Split the dataset 90% train, 10% test
    train_test_split = full_dataset.train_test_split(test_size=0.1, shuffle=True, seed=42)
    
    datasets = DatasetDict({
        'train': train_test_split['train'],
        'test': train_test_split['test']
    })
    print("Dataset loaded and split successfully.")
    print(datasets)
except Exception as e:
    print(f"Error loading dataset: {e}")
    raise

# --- 3. Load Tokenizer & Preprocessing ---
print(f"Loading T5 tokenizer from local path: {MODEL_CHECKPOINT}")
tokenizer = AutoTokenizer.from_pretrained(MODEL_CHECKPOINT)

# We need to format the input for T5
prefix = "question: "
def preprocess_function_t5(examples):
    # Inputs: "question: [QUESTION] context: [CONTEXT]"
    inputs = [
        f"{prefix}{q} context: {c}" 
        for q, c in zip(examples["generated_question"], examples["context"])
    ]
    
    # Outputs: The answer text
    targets = [ans for ans in examples["generated_answer"]]
    
    # Tokenize inputs and targets
    model_inputs = tokenizer(inputs, max_length=512, truncation=True, padding="max_length")
    labels = tokenizer(targets, max_length=128, truncation=True, padding="max_length")

    # Set the tokenized targets as the 'labels'
    model_inputs["labels"] = labels["input_ids"]
    return model_inputs

print("Applying T5 preprocessing to the dataset...")
tokenized_datasets = datasets.map(preprocess_function_t5, batched=True)
print("Preprocessing complete.")

# --- 4. Fine-Tuning ---
print(f"Loading T5 model from local path: {MODEL_CHECKPOINT}")
model = AutoModelForSeq2SeqLM.from_pretrained(MODEL_CHECKPOINT)

# Use DataCollatorForSeq2Seq for generative models
data_collator = DataCollatorForSeq2Seq(tokenizer=tokenizer, model=model)

# Set Training Arguments
training_args = TrainingArguments(
    output_dir=NEW_MODEL_DIR,
    eval_strategy="epoch",
    logging_strategy="epoch", # So you can see the training loss
    learning_rate=2e-5,
    per_device_train_batch_size=8, # T5 is bigger, use a smaller batch size
    per_device_eval_batch_size=8,
    num_train_epochs=3,
    weight_decay=0.01,
    push_to_hub=False,
    save_strategy="epoch",
    load_best_model_at_end=True,
    report_to="none",
)

# Initialize Trainer
print("Initializing Trainer...")
trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=tokenized_datasets["train"],
    eval_dataset=tokenized_datasets["test"],
    tokenizer=tokenizer,
    data_collator=data_collator,
)

print("\n🚀 *** STARTING T5 FINE-TUNING! *** 🚀")
try:
    trainer.train()
    print("\n✅ *** T5 FINE-TUNING COMPLETE! *** ✅")
    print(f"Saving the best model to {NEW_MODEL_DIR}...")
    trainer.save_model()
    print("Model saved successfully.")
except torch.cuda.OutOfMemoryError:
    print("\n❌ *** CUDA Out of Memory! *** ❌")
    print("Your GPU ran out of memory. Try this:")
    print("1. Restart the kernel.")
    print("2. In this cell, change 'per_device_train_batch_size=8' to 'per_device_train_batch_size=4' (and eval to 4).")
    print("3. Re-run this cell.")
except Exception as e:
    print(f"\nAn error occurred during training: {e}")
```
